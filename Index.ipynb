# **Index**

## **What is Index?**

- A database index is an **auxiliary** data structure that organizes **key values** and associates them with the relevant **table rows**, allowing the query engine to locate data efficiently while trading <span style = "color:red">**additional storage**</span> and <span style = "color:red">**write-maintenance**</span> overhead for improved <span style = "color:green">**read performance**</span>.
- Storing data without an effective index is like throwing papers into a drawer without sorting them: it’s quick to put them away (<span style = "color:green">**WRITE**</span>), but slow to find the one you need (<span style = "color:red">**READ**</span>).
- Data structure provides quick access to your data, optimizing the speed of your queries.
- <span style = "color:orange">**Key characteristics:**</span>
    - **Purpose:** Improve data retrieval, filtering, joins, and sometimes sorting or grouping.
    - **Structure:** Stores index key values in an organized structure with row references or, for a clustered index, the actual data rows at the leaf level.
    - **Trade-off:** Requires additional storage and maintenance during data modifications.
    - **Limitation:** An index does not guarantee faster execution; the query optimizer chooses whether using it is beneficial.


## **Index Category**

- <span style = "color:Orange">**Structure:**</span> How the database is organizing and referencing the data.
    - Clustered Index, Non-Clustered Index
- <span style = "color:Orange">**Storage:**</span> How the data is stored physically in database.
    - Rowstore Index, Columnstore Index
- <span style = "color:Orange">**Function:**</span>
    - Unique Index, Filtered Index

## **Database Page (SQL Server)**

- A database page is an 8-KB (8,192 bytes) storage structure containing a page header, data or index records (depending on page type), and page-management information. SQL Server organizes pages into extents and allocates them to database objects such as tables and indexes.
- **Page Structure:**
    - Page header — 96 bytes
        - Page type, page ID, object/allocation metadata, free-space information

    - Data records
        - Rows stored on the page

    - Row offset array
        - 2-byte entry per row, identifying row positions
- **Page Types:** Data page, Index page
- <span style = "color:purple">**Data file (.mdf) = ∑ (Data pages, Index pages)**</span>

- HEAP = a table without a Clustered Index, Fast WRITE and Slow READ
    - Analogy: throwing a paper to your drawer without sorting --> fast to put but slow to find it.
